```python
import re
import pandas as pd
import numpy as np
from collections import Counter
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from sklearn.metrics import accuracy_score, classification_report
```

```python
import random
# Set random seeds for reproducibility
random.seed(42)
torch.manual_seed(42)
```

```python
# ==========================================
# 1. DATA EXTRACTION & CLEANING
# ==========================================

def clean_and_tokenize(text):
    """Cleans punctuation, URLs, tickers, handles lowercase conversion and splits into words."""
    text = text.lower()
    text = re.sub(r'https?://\S+|www\.\S+', '', text)  # Strip URLs
    text = re.sub(r'\$\w+', '', text)                  # Strip Stock Tickers (e.g. $AAPL)
    text = re.sub(r'[^a-z0-9\s%+-]', '', text)         # Keep relevant finance syntax characters
    tokens = text.strip().split()
    return tokens
```

```python
raw_dataset=train
for text, label in enumerate(raw_dataset):
    print(text)
print(raw_dataset.columns)
print(raw_dataset)
```

```python
# Clean all tweets
tokenized_data = [
    (clean_and_tokenize(text), label)
    for text, label in zip(raw_dataset["text"], raw_dataset["label"])
]
```

```python
# ==========================================
# 2. VOCABULARY GENERATION FROM SCRATCH
# ==========================================
vocab = {"<PAD>": 0, "<UNK>": 1}
for tokens, _ in tokenized_data:
    for token in tokens:
        if token not in vocab:
            vocab[token] = len(vocab)

VOCAB_SIZE = len(vocab)
MAX_LEN = max(len(tokens) for tokens, _ in tokenized_data)
```

```python
def numericalize_and_pad(tokens, vocab, max_len):
    """Converts token sequences to integer indices and pads sequences to maximum dataset length."""
    indices = [vocab.get(token, vocab["<UNK>"]) for token in tokens]
    if len(indices) < max_len:
        indices += [vocab["<PAD>"]] * (max_len - len(indices))
    else:
        indices = indices[:max_len]
    return indices
```

```python
# Convert full text dataset to integer tensors
X = torch.tensor([numericalize_and_pad(tokens, vocab, MAX_LEN) for tokens, _ in tokenized_data], dtype=torch.long)
y = torch.tensor([label for _, label in tokenized_data], dtype=torch.long)

# Shuffle and split into Train (80%) and Validation (20%) datasets
dataset_size = len(raw_dataset)
indices = list(range(dataset_size))
random.shuffle(indices)

split = int(dataset_size * 0.8)
train_idx, val_indices = indices[:split], indices[split:]

X_train, y_train = X[train_idx], y[train_idx]
X_val, y_val = X[val_indices], y[val_indices]
```

```python
import torch.nn.functional as F
```

```python
# ==========================================
# 3. DEEP LEARNING MODEL ARCHITECTURE
# ==========================================
class FinancialTweetClassifier(nn.Module):
    def __init__(self, vocab_size, embedding_dim, hidden_dim, output_dim=3):
        super(FinancialTweetClassifier, self).__init__()
        # Embedding Layer mapping index vectors to high-dimensional space
        self.embedding = nn.Embedding(vocab_size, embedding_dim, padding_idx=0)
        
        # Linear hidden conversion layer (Acting as simple feature aggregator)
        self.fc1 = nn.Linear(embedding_dim, hidden_dim)
        self.dropout = nn.Dropout(0.3)
        
        # Classification Output Layer mapping to logits over 3 Sentiment Scales
        self.fc2 = nn.Linear(hidden_dim, output_dim)
        
    def forward(self, text):
        # text dimensions: [batch_size, seq_len]
        embedded = self.embedding(text) # [batch_size, seq_len, embedding_dim]
        
        # Compute dynamic document representation via temporal Global Average Pooling
        pooled = torch.mean(embedded, dim=1) # [batch_size, embedding_dim]
        
        # Dense feature conversion tracking multi-class signals
        hidden = F.relu(self.fc1(pooled))
        hidden = self.dropout(hidden)
        logits = self.fc2(hidden) # [batch_size, output_dim]
        return logits

```

```python
# Initialize hyper-parameters
EMBEDDING_DIM = 32
HIDDEN_DIM = 16
BATCH_SIZE = 4
EPOCHS = 15

model = FinancialTweetClassifier(vocab_size=VOCAB_SIZE, embedding_dim=EMBEDDING_DIM, hidden_dim=HIDDEN_DIM)
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.01)
```

```python
# ==========================================
# 4. DATA BATCHING & TRAINING LOOP
# ==========================================
def get_batches(X_data, y_data, batch_size):
    """Yields programmatic iterable slices matching processing constraints manually."""
    num_samples = len(X_data)
    for i in range(0, num_samples, batch_size):
        yield X_data[i : i + batch_size], y_data[i : i + batch_size]

print("Starting Sentiment Engine Training...\n" + "="*40)

for epoch in range(1, EPOCHS + 1):
    model.train()
    epoch_loss = 0.0
    
    # Process sequential training batch segments
    for batch_X, batch_y in get_batches(X_train, y_train, BATCH_SIZE):
        optimizer.zero_grad()
        predictions = model(batch_X)
        loss = criterion(predictions, batch_y)
        loss.backward()
        optimizer.step()
        epoch_loss += loss.item() * len(batch_y)
        
    train_loss = epoch_loss / len(X_train)
    
    # Track cross-entropy validation loss
    model.eval()
    with torch.no_grad():
        val_preds = model(X_val)
        val_loss = criterion(val_preds, y_val).item()
        val_acc = (val_preds.argmax(dim=1) == y_val).float().mean().item() * 100
        
    print(f"Epoch {epoch:02d}/{EPOCHS} | Train Loss: {train_loss:.4f} | Val Loss: {val_loss:.4f} | Val Accuracy: {val_acc:.1f}%")

print("="*40 + "\nModel successfully optimized.")
```

```python
# ==========================================
# 5. INFERENCE IN ENGINE STRATEGY UTILITY
# ==========================================
def predict_tweet_sentiment(tweet, model, vocab, max_len):
    """Processes, sequence maps, indices, and evaluates arbitrary sample text input."""
    label_map = {0: "Bearish 📉", 1: "Bullish 📈", 2: "Neutral ⚖️"}
    
    model.eval()
    with torch.no_grad():
        tokens = clean_and_tokenize(tweet)
        padded_sequence = numericalize_and_pad(tokens, vocab, max_len)
        input_tensor = torch.tensor([padded_sequence], dtype=torch.long)
        
        logits = model(input_tensor)
        probabilities = F.softmax(logits, dim=1)
        predicted_class = torch.argmax(probabilities, dim=1).item()
        
    return label_map[predicted_class], probabilities[0].tolist()

# Test the system with an outside phrase instance
test_tweet = "Analysts upgrade targets after blowout record revenue metrics!"
sentiment, distributions = predict_tweet_sentiment(test_tweet, model, vocab, MAX_LEN)

print(f"\nEvaluating Input Phrase: '{test_tweet}'")
print(f"Predicted Sentiment Domain: {sentiment}")
print(f"Network Class Probabilities -> Bearish: {distributions[0]:.2%}, Bullish: {distributions[1]:.2%}, Neutral: {distributions[2]:.2%}")
```
